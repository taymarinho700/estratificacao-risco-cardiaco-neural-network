```python
# ==============================================================================
# SEÇÃO 1: CARGA E PRÉ-PROCESSAMENTO COMPLETO
# ==============================================================================
# 1. Carga dos dados
df = pd.read_csv('synthetic_heart_disease_dataset.csv')

# 2. Tratamento de Valores Ausentes (NaN) por Tipo de Dado
# Preenche colunas de texto com 'Unknown'
cols_texto = df.select_dtypes(include=['object', 'category', 'string']).columns
df[cols_texto] = df[cols_texto].fillna('Unknown')

# Preenche colunas numéricas com a mediana
cols_numericas = df.select_dtypes(include=['number']).columns
df[cols_numericas] = df[cols_numericas].fillna(df[cols_numericas].median())

# 3. Codificação de Variáveis Categóricas (One-Hot Encoding)
df_encoded = pd.get_dummies(df, drop_first=True)

# 4. Separação das Features (X) e do Alvo (y)
X = df_encoded.drop(columns=['Heart_Disease']).values
y = df_encoded['Heart_Disease'].values
feature_names = df_encoded.drop(columns=['Heart_Disease']).columns

# 5. Padronização de Escala (StandardScaler)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 6. Divisão Treino (75%) e Teste (25%) com Estratificação
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.25, random_state=42, stratify=y
)

print("=== PRÉ-PROCESSAMENTO CONCLUÍDO COM SUCESSO ===")
print(f"Total de Atributos após Codificação: {X_train.shape[1]}")
print(f"Amostras de Treino: {X_train.shape[0]} | Amostras de Teste: {X_test.shape[0]}")
```

```python
# ==============================================================================
# SEÇÃO 2: CLASSE REDE NEURAL "FROM SCRATCH"
# ==============================================================================

class AlgoritmoNeuralNetwork:
    """
    Modelo Perceptron/Regressão Logística construído do zero com NumPy.
    Desenvolvido com foco em estabilidade numérica e calibração preventiva.
    """
    def __init__(self, taxa_aprendizado=0.01, num_interacoes=2000): 
        self.taxa_aprendizado = taxa_aprendizado
        self.num_interacoes = num_interacoes
        self.pesos = None
        self.bias = None

    def func_activation_sigmoid(self, pred): 
        # Estabilidade numérica (limita valores para evitar overflow)
        pred_safe = np.clip(pred, -500, 500)
        return 1 / (1 + np.exp(-pred_safe))

    def fit(self, X, y):
        num_registros, num_atributos = X.shape

        # Inicialização dinâmica dos parâmetros baseada no número de colunas (26)
        self.pesos = np.zeros(num_atributos)
        self.bias = 0.0

        print("--> Treinamento Inicializado...")

        for i in range(self.num_interacoes): 
            # 1. Forward Pass
            previsao_bruta = np.dot(X, self.pesos) + self.bias 
            previsao_final = self.func_activation_sigmoid(previsao_bruta)

            # 2. Cálculo do Erro
            erro = previsao_final - y

            # 3. Backward Pass (Gradientes)
            dw = (1 / num_registros) * np.dot(X.T, erro)
            db = (1 / num_registros) * np.sum(erro)

            # 4. Atualização dos Parâmetros
            self.pesos -= self.taxa_aprendizado * dw
            self.bias -= self.taxa_aprendizado * db

            # Log de progresso a cada 20% do total de iterações
            if (i + 1) % (self.num_interacoes // 5) == 0 or i == 0:
                loss = -np.mean(y * np.log(previsao_final + 1e-15) + (1 - y) * np.log(1 - previsao_final + 1e-15))
                print(f"    Iteração {i+1}/{self.num_interacoes} - Perda (Log-Loss): {loss:.4f}")

        print("--> Treinamento Concluído!")

    def predict_proba(self, X):
        # Retorna a probabilidade contínua (entre 0 e 1)
        previsao_bruta = np.dot(X, self.pesos) + self.bias
        return self.func_activation_sigmoid(previsao_bruta)

    def predict(self, X, limiar=0.35):
        # Aplica o limiar clínico customizável (padrão em 35% para sensibilidade)
        probabilidades = self.predict_proba(X)
        return (probabilidades >= limiar).astype(int)
```

```python
# ==============================================================================
# SEÇÃO 3: TREINAMENTO, AVALIAÇÃO MÉDICA E INTERPRETAÇÃO
# ==============================================================================

from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score

# 1. Instanciando e Treinando o Modelo Customizado
modelo = AlgoritmoNeuralNetwork(taxa_aprendizado=0.05, num_interacoes=1000)
modelo.fit(X_train, y_train)

# 2. Gerando Previsões no Conjunto de Teste (12.500 registros)
probs_teste = modelo.predict_proba(X_test)
previsoes_teste = modelo.predict(X_test, limiar=0.35)

# 3. Exibição das Métricas Clínicas Profissionais
print("\n" + "="*60)
print("RELATÓRIO DE DESEMPENHO MÉDICO (LIMIAR CLÍNICO = 35%)")
print("="*60)
print(classification_report(y_test, previsoes_teste, target_names=['Saudável (0)', 'Risco Cardíaco (1)']))

roc_auc = roc_auc_score(y_test, probs_teste)
print(f"Pontuação ROC-AUC: {roc_auc:.4f}")

# 4. EXPLAINABILITY: Mapeamento do Impacto dos Exames nos Pesos
df_pesos = pd.DataFrame({
    'Exame / Atributo': feature_names,
    'Peso Aprendido (Coeficiente)': modelo.pesos
}).sort_values(by='Peso Aprendido (Coeficiente)', ascending=False)

print("\n" + "="*60)
print("IMPACTO DOS ATRIBUTOS NO DIAGNÓSTICO DE RISCO CARDÍACO")
print("="*60)
print(df_pesos.to_string(index=False))
```

```python
# ==============================================================================
# SEÇÃO 4: VISUALIZAÇÃO GRÁFICA DOS RESULTADOS (SEM AVISOS)
# ==============================================================================

import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 1. Matriz de Confusão
cm = confusion_matrix(y_test, previsoes_teste)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['Saudável', 'Risco Cardíaco'],
            yticklabels=['Saudável', 'Risco Cardíaco'])
axes[0].set_title('Matriz de Confusão (Limiar = 35%)')
axes[0].set_ylabel('Classe Real')
axes[0].set_xlabel('Classe Predita pelo Modelo')

# 2. Gráfico dos Top 8 Pesos Mais Impactantes (Sintaxe Atualizada para Seaborn v0.14+)
df_top = pd.concat([df_pesos.head(5), df_pesos.tail(3)])
sns.barplot(
    data=df_top, 
    x='Peso Aprendido (Coeficiente)', 
    y='Exame / Atributo', 
    hue='Exame / Atributo', 
    palette='coolwarm', 
    legend=False, 
    ax=axes[1]
)
axes[1].set_title('Top Fatores de Impacto no Modelo')
axes[1].set_xlabel('Magnitude do Peso (Coeficiente Logístico)')

plt.tight_layout()
plt.show()
```
